```python
import pandas as pd

# Bước 1: Tạo 2 bảng phụ để mapping từng điểm với khoảng cách
df_a = df[['Mã phòng ban 1', 'Khoảng cách đường bộ (km)']].rename(columns={
    'Mã phòng ban 1': 'Mã phòng ban'
})
df_all = pd.concat([df_a, df_b], axis=0)

# Bước 2: Lấy khoảng cách nhỏ nhất tương ứng với mỗi phòng ban
min_kc = df_all.groupby('Mã phòng ban')['Khoảng cách đường bộ (km)'].min().reset_index()
min_kc = min_kc.rename(columns={'Khoảng cách đường bộ (km)': 'MinDistance'})

# Bước 3: Gắn vào bảng gốc và lọc
# Với phòng ban 1
df_1 = df.merge(min_kc, left_on='Mã phòng ban 1', right_on='Mã phòng ban')
mask_1 = df_1['Khoảng cách đường bộ (km)'] == df_1['MinDistance']

# Với phòng ban 2
df_2 = df.merge(min_kc, left_on='Mã phòng ban 2', right_on='Mã phòng ban')
mask_2 = df_2['Khoảng cách đường bộ (km)'] == df_2['MinDistance']

# Kết hợp các dòng thỏa mãn điều kiện từ cả A và B
df_filtered = pd.concat([df_1[mask_1], df_2[mask_2]])

# Bỏ cột phụ 'MinDistance', 'Mã phòng ban' (vì từ merge)
df_filtered = df_filtered[df.columns]
```
