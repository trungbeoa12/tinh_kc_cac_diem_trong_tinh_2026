```python
import unicodedata
import re

def normalize_column_name(col):
    # chuyển Đ đ trước khi remove accent
    col = col.replace("Đ", "D").replace("đ", "d")
    
    # bỏ dấu tiếng Việt
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    
    # lower
    col = col.lower()
    
    # bỏ khoảng trắng đầu cuối
    col = col.strip()
    
    # thay khoảng trắng bằng _
    col = re.sub(r"\s+", "_", col)
    
    # thay . bằng _
    col = col.replace(".", "_")
    
    # bỏ ký tự đặc biệt
    col = re.sub(r"[^a-z0-9_]", "", col)
    
    # bỏ _ trùng nhau
    col = re.sub(r"_+", "_", col)
    
    return col
```

```python
import pandas as pd
from itertools import permutations
from math import radians, sin, cos, sqrt, atan2

# Hàm tính khoảng cách chim bay (Haversine)
def haversine(lon1, lat1, lon2, lat2):
    R = 6371  # bán kính trái đất (km)

    lon1, lat1, lon2, lat2 = map(radians, [lon1, lat1, lon2, lat2])

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    return R * c


result = []

for tinh_thanh, group in ml_2026.groupby("tinh_thanh"):
    
    records = group[["ma_phong_ban", "ten_phong_ban", "kinh_do", "vi_do"]].to_dict("records")

    # permutations để lấy cả AB và BA
    for r1, r2 in permutations(records, 2):

        khoang_cach = haversine(
            r1["kinh_do"], r1["vi_do"],
            r2["kinh_do"], r2["vi_do"]
        )

        result.append({
            "ma_phong_ban_1": r1["ma_phong_ban"],
            "ten_phong_ban_1": r1["ten_phong_ban"],
            "kinh_do_1": r1["kinh_do"],
            "vi_do_1": r1["vi_do"],
            "ma_phong_ban_2": r2["ma_phong_ban"],
            "ten_phong_ban_2": r2["ten_phong_ban"],
            "kinh_do_2": r2["kinh_do"],
            "vi_do_2": r2["vi_do"],
            "khoang_cach_chim_bay": khoang_cach,
            "tinh_thanh": tinh_thanh
        })

ml_2026_to_hop = pd.DataFrame(result)
ml_2026_to_hop['ma_phong_ban_1'] = ml_2026_to_hop['ma_phong_ban_1'].astype('int64')
ml_2026_to_hop['ma_phong_ban_2'] = ml_2026_to_hop['ma_phong_ban_2'].astype('int64')
```

```python
import pandas as pd
import re

def tach_khoang_cach_va_thoi_gian(text):
    if pd.isna(text) or "Không tìm thấy" in str(text):
        return pd.Series([None, None])
    
    text = str(text).replace('\n', ' ').strip().lower()

    # Trường hợp: "1 hr 27 min 36.6 km"
    match = re.search(r'(?:(\d+)\s*hr)?\s*(\d+)?\s*min.*?([\d.]+)\s*(km|m)', text)
    if match:
        hours = int(match.group(1)) if match.group(1) else 0
        mins = int(match.group(2)) if match.group(2) else 0
        distance = float(match.group(3))
        unit = match.group(4)
        total_mins = hours * 60 + mins
        distance_km = distance / 1000 if unit == 'm' else distance
        return pd.Series([total_mins, distance_km])

    # Trường hợp: "2 min 750 m" (không có hr)
    match = re.search(r'(\d+)\s*min.*?([\d.]+)\s*(km|m)', text)
    if match:
        mins = int(match.group(1))
        distance = float(match.group(2))
        unit = match.group(3)
        distance_km = distance / 1000 if unit == 'm' else distance
        return pd.Series([mins, distance_km])

    # Trường hợp chỉ có "1 hr 48.9 km"
    match = re.search(r'(\d+)\s*hr.*?([\d.]+)\s*(km|m)', text)
    if match:
        hours = int(match.group(1))
        distance = float(match.group(2))
        unit = match.group(3)
        distance_km = distance / 1000 if unit == 'm' else distance
        return pd.Series([hours * 60, distance_km])

    # Trường hợp chỉ có khoảng cách, ví dụ: "600 m" hoặc "31.8 km"
    match = re.search(r'([\d.]+)\s*(km|m)', text)
    if match:
        distance = float(match.group(1))
        unit = match.group(2)
        distance_km = distance / 1000 if unit == 'm' else distance
        return pd.Series([None, distance_km])

    return pd.Series([None, None])
```

```python
df = df_final.copy()

# bỏ self-pair nếu có
df = df[df['ma_phong_ban_1'] != df['ma_phong_ban_2']].copy()

# cờ cùng đơn vị
df['cung_don_vi'] = df['ma_don_vi_1'] == df['ma_don_vi_2']

# hàm gộp tên, bỏ trùng, nối bằng ;
def gop_ten(series):
    vals = (
        series.dropna()
        .astype(str)
        .str.strip()
    )
    vals = vals[vals != '']
    vals = pd.unique(vals)
    return '; '.join(vals)

# 1) Phongban_0.5km
pb_05 = (
    df[df['khoang_cach_duong_bo_km'] <= 0.5]
    .groupby('ma_phong_ban_1')['ten_phong_ban_2']
    .apply(gop_ten)
    .rename('Phongban_0.5km')
)

# 2) Phongban_1km
pb_1 = (
    df[df['khoang_cach_duong_bo_km'] <= 1]
    .groupby('ma_phong_ban_1')['ten_phong_ban_2']
    .apply(gop_ten)
    .rename('Phongban_1km')
)

# 3) phongban_1km (cột mới)
pb_1_lower = (
    df[df['khoang_cach_duong_bo_km'] <= 1]
    .groupby('ma_phong_ban_1')['ten_phong_ban_2']
    .apply(gop_ten)
    .rename('phongban_1km')
)

# 4) Phongban_0.5km_cung_donvi
pb_05_dv = (
    df[(df['khoang_cach_duong_bo_km'] <= 0.5) & (df['cung_don_vi'])]
    .groupby('ma_phong_ban_1')['ten_phong_ban_2']
    .apply(gop_ten)
    .rename('Phongban_0.5km_cung_donvi')
)

# 5) Phongban_1km_cung_donvi
pb_1_dv = (
    df[(df['khoang_cach_duong_bo_km'] <= 1) & (df['cung_don_vi'])]
    .groupby('ma_phong_ban_1')['ten_phong_ban_2']
    .apply(gop_ten)
    .rename('Phongban_1km_cung_donvi')
)

# merge vào bảng kết quả
df_ket_qua = (
    df_ket_qua
    .merge(pb_05, on='ma_phong_ban_1', how='left')
    .merge(pb_1, on='ma_phong_ban_1', how='left')
    .merge(pb_1_lower, on='ma_phong_ban_1', how='left')   # cột mới
    .merge(pb_05_dv, on='ma_phong_ban_1', how='left')
    .merge(pb_1_dv, on='ma_phong_ban_1', how='left')
)

# fill rỗng
cols_text = [
    'Phongban_0.5km',
    'Phongban_1km',
    'phongban_1km',
    'Phongban_0.5km_cung_donvi',
    'Phongban_1km_cung_donvi'
]

df_ket_qua[cols_text] = df_ket_qua[cols_text].fillna('')
```

```python
import pandas as pd
import numpy as np

df = df_final_2.copy()

# =========================================
# 1. Chuẩn hóa dữ liệu
# =========================================
num_cols = [
    'kinh_do_1', 'vi_do_1',
    'kinh_do_2', 'vi_do_2',
    'khoang_cach_duong_bo_km',
    'khoang_cach_chim_bay_km'
]

for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

text_cols = [
    'ma_phong_ban_1', 'ten_phong_ban_1',
    'ma_phong_ban_2', 'ten_phong_ban_2',
    'ma_don_vi_1', 'ma_don_vi_2',
    'Tinh_thanh(cu)', 'Tinh_thanh(moi)'
]

for col in text_cols:
    df[col] = df[col].astype(str).str.strip()

# bỏ self-pair: A -> A
df = df[df['ma_phong_ban_1'] != df['ma_phong_ban_2']].copy()

# =========================================
# 2. Base table: 1 dòng / điểm giao dịch gốc
# =========================================
base = (
    df[
        [
            'ma_phong_ban_1', 'ten_phong_ban_1',
            'kinh_do_1', 'vi_do_1',
            'ma_don_vi_1',
            'Tinh_thanh(cu)', 'Tinh_thanh(moi)'
        ]
    ]
    .drop_duplicates(subset=['ma_phong_ban_1'])
    .copy()
)

# =========================================
# 3. Hàm lấy điểm gần nhất
# =========================================
def lay_diem_gan_nhat(df_in, distance_col, prefix):
    """
    df_in: dataframe đã lọc đúng phạm vi cần xét
    distance_col: cột khoảng cách để tìm min
    prefix: tiền tố tên cột output
    """
    tmp = df_in.dropna(subset=[distance_col]).copy()
    tmp = tmp.sort_values(['ma_phong_ban_1', distance_col, 'ma_phong_ban_2'])

    nearest = (
        tmp.groupby('ma_phong_ban_1', as_index=False)
           .first()[['ma_phong_ban_1', 'ma_phong_ban_2', 'ten_phong_ban_2', distance_col]]
           .rename(columns={
               'ma_phong_ban_2': f'{prefix}_ma_pgd',
               'ten_phong_ban_2': f'{prefix}_ten_pgd',
               distance_col: f'{prefix}_khoang_cach'
           })
    )
    return nearest

# =========================================
# 4. Gần nhất trong tỉnh cũ - đường chim bay
# =========================================
# Vì mỗi dòng đang là cặp trong cùng tỉnh cũ/tỉnh mới tương ứng của điểm 1,
# nếu dữ liệu của bạn đã được tạo theo logic cùng tỉnh thì có thể dùng trực tiếp.
# Nếu muốn chặt chẽ hơn thì chỉ cần giữ nguyên từng tỉnh cũ của điểm 1.
df_tinh_cu_cb = df.copy()

nearest_tinh_cu_cb = lay_diem_gan_nhat(
    df_in=df_tinh_cu_cb,
    distance_col='khoang_cach_chim_bay_km',
    prefix='tinh_cu_chim_bay'
)

# =========================================
# 5. Gần nhất trong tỉnh cũ - đường thực tế
# =========================================
df_tinh_cu_db = df.copy()

nearest_tinh_cu_db = lay_diem_gan_nhat(
    df_in=df_tinh_cu_db,
    distance_col='khoang_cach_duong_bo_km',
    prefix='tinh_cu_duong_bo'
)

# =========================================
# 6. Gần nhất trong CN - đường chim bay
# =========================================
df_cn_cb = df[df['ma_don_vi_1'] == df['ma_don_vi_2']].copy()

nearest_cn_cb = lay_diem_gan_nhat(
    df_in=df_cn_cb,
    distance_col='khoang_cach_chim_bay_km',
    prefix='cn_chim_bay'
)

# =========================================
# 7. Gần nhất trong CN - đường thực tế
# =========================================
df_cn_db = df[df['ma_don_vi_1'] == df['ma_don_vi_2']].copy()

nearest_cn_db = lay_diem_gan_nhat(
    df_in=df_cn_db,
    distance_col='khoang_cach_duong_bo_km',
    prefix='cn_duong_bo'
)

# =========================================
# 8. Gần nhất trong tỉnh mới - đường thực tế
# =========================================
# Nếu toàn bộ cặp đã cùng Tinh_thanh(moi) thì dùng trực tiếp.
# Nếu không, cần có thêm cột tỉnh mới của điểm 2 để lọc chặt chẽ hơn.
df_tinh_moi_db = df.copy()

nearest_tinh_moi_db = lay_diem_gan_nhat(
    df_in=df_tinh_moi_db,
    distance_col='khoang_cach_duong_bo_km',
    prefix='tinh_moi_duong_bo'
)

# =========================================
# 9. Merge toàn bộ vào base
# =========================================
result = (
    base
    .merge(nearest_tinh_cu_cb, on='ma_phong_ban_1', how='left')
    .merge(nearest_tinh_cu_db, on='ma_phong_ban_1', how='left')
    .merge(nearest_cn_cb, on='ma_phong_ban_1', how='left')
    .merge(nearest_cn_db, on='ma_phong_ban_1', how='left')
    .merge(nearest_tinh_moi_db, on='ma_phong_ban_1', how='left')
)

# =========================================
# 10. Rename cho giống logic trong file Excel
# =========================================
result = result.rename(columns={
    'ma_phong_ban_1': 'Mã PGD gốc',
    'ten_phong_ban_1': 'Tên PGD gốc',
    'kinh_do_1': 'Kinh độ',
    'vi_do_1': 'Vĩ độ',
    'ma_don_vi_1': 'Mã đơn vị',
    'Tinh_thanh(cu)': 'Tỉnh thành cũ',
    'Tinh_thanh(moi)': 'Tỉnh thành mới',

    'tinh_cu_chim_bay_ma_pgd': 'Mã PGD_gan_nhat_tinh_cu_chim_bay',
    'tinh_cu_chim_bay_ten_pgd': 'Tên PGD_gan_nhat_tinh_cu_chim_bay',
    'tinh_cu_chim_bay_khoang_cach': 'Khoảng cách_tinh_cu_chim_bay',

    'tinh_cu_duong_bo_ma_pgd': 'Mã PGD_gan_nhat_tinh_cu_duong_bo',
    'tinh_cu_duong_bo_ten_pgd': 'Tên PGD_gan_nhat_tinh_cu_duong_bo',
    'tinh_cu_duong_bo_khoang_cach': 'Khoảng cách_tinh_cu_duong_bo',

    'cn_chim_bay_ma_pgd': 'Mã PGD_gan_nhat_CN_chim_bay',
    'cn_chim_bay_ten_pgd': 'Tên PGD_gan_nhat_CN_chim_bay',
    'cn_chim_bay_khoang_cach': 'Khoảng cách_CN_chim_bay',

    'cn_duong_bo_ma_pgd': 'Mã PGD_gan_nhat_CN_duong_bo',
    'cn_duong_bo_ten_pgd': 'Tên PGD_gan_nhat_CN_duong_bo',
    'cn_duong_bo_khoang_cach': 'Khoảng cách_CN_duong_bo',

    'tinh_moi_duong_bo_ma_pgd': 'Mã PGD_gan_nhat_tinh_moi_duong_bo',
    'tinh_moi_duong_bo_ten_pgd': 'Tên PGD_gan_nhat_tinh_moi_duong_bo',
    'tinh_moi_duong_bo_khoang_cach': 'Khoảng cách_tinh_moi_duong_bo',
})

```
