```python
kc = pd.read_excel(r'/home/trungdt2/Documents/GIS_VTB_Project2025/crawl_all_khoang_cach/etl_20250604/full_trong_tinh_sau_sat_nhap_20250520.xlsx')
```

```python
df_ma_tinh_cu = df[['Mã phòng ban','Tỉnh thành (hiện tại)']]
```

```python
df_ma_tinh_cu.head()
```

```python
kc = pd.merge(kc,df_ma_tinh_cu,left_on='Mã phòng ban 1',right_on='Mã phòng ban',how='left')
del kc['Mã phòng ban']
```

```python
kc.head()
```

```python
kc = pd.merge(kc,df_ma_tinh_cu,left_on='Mã phòng ban 2',right_on='Mã phòng ban',how='left')
del kc['Mã phòng ban']
```

```python
kc[kc['Tỉnh thành (hiện tại)_x']=='Thái Bình']
```

```python
# Lọc các dòng có giá trị của 'Tỉnh thành (hiện tại)_x' bằng 'Tỉnh thành (hiện tại)_y'
filtered_kc = kc[kc['Tỉnh thành (hiện tại)_x'] == kc['Tỉnh thành (hiện tại)_y']]
```

```python
filtered_kc.head()
```

```python
# Lọc ra các dòng có "Khoảng cách đường bộ (km)" nhỏ nhất theo "Mã phòng ban 1"
df_min_distance = filtered_kc[
    filtered_kc["Khoảng cách đường bộ (km)"] == 
    filtered_kc.groupby("Mã phòng ban 1")["Khoảng cách đường bộ (km)"].transform('min')
]
```

```python
df_min_distance.head()
```

```python
del df_min_distance['Tỉnh thành (sau sáp nhập)']
del df_min_distance['Tỉnh thành (hiện tại)_x']
```

```python
df_min_distance.to_excel(r'/home/trungdt2/Documents/GIS_VTB_Project2025/crawl_all_khoang_cach/etl_20250604/tinh_cu_20250604.xlsx')
```
